```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('Dataset.csv')
df.head()
```

```python
print(df.shape)
print(df.dtypes)
```

```python
X = np.array(df[['age', 'sex', 'cp', 'trtbps', 'chol', 'fbs', 'restecg', 'thalachh', 'exng']])
X[:5]
```

```python
y = np.array(df[['output']])
y[:5]
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=3
)
```

```python
print(x_train.shape, end=' ')
print(y_train.shape)
print(x_test.shape, end=' ')
print(y_test.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

criterions = ['gini', 'entropy', 'log_loss']
for i in criterions:
    model = DecisionTreeClassifier(criterion=i, random_state=3)
    model.fit(x_train, y_train)
    y_pred = model.predict(x_test)
    print(f'criterion is {i} and its accuracy score : {accuracy_score(y_test, y_pred)}')
```
