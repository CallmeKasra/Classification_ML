```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('Dataset.csv')
df.sample(5, random_state=12)
```

```python
X = np.array(df[['age', 'sex', 'cp', 'trtbps', 'chol', 'fbs', 'restecg', 'thalachh', 'exng']])
y = np.array(df['output'])
print(f'x :\n{X[:5]}\ny :\n{y[:5]}')
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=12
)
```

```python
print(x_train.shape, end=' ')
print(y_train.shape)
print(x_test.shape, end=' ')
print(y_test.shape)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

accuracy = []
for n in range(1, 10):
    model = KNeighborsClassifier(n_neighbors=n)
    model.fit(x_train, y_train)
    y_hat = model.predict(x_test)
    accuracy.append(accuracy_score(y_test, y_hat))
    
result = pd.DataFrame({'K':range(1, 10), 'Accuracy':accuracy})
result
```

```python
model = KNeighborsClassifier(n_neighbors=9, weights='distance')
model.fit(x_train, y_train)
y_hat = model.predict(x_test)
print(f'accuracy with k(9) and weights(distance) is = {accuracy_score(y_test, y_hat)}')
```
